```python
# Writing data to a file

file = open("student.txt", "w")

file.write("Name: John\n")
file.write("Roll Number: 100\n")
file.write("Marks: 85\n")

file.close()

print("Data written successfully.")
```

```python
# Reading data from a file

file = open("student.txt", "r")

data = file.read()

print(data)

file.close()
```

```python
# Writing to a file

with open("student.txt", "w") as file:
    file.write("Name: Rahul\n")
    file.write("Marks: 85\n")

# Reading from the file

with open("student.txt", "r") as file:
    data = file.read()

print(data)

with open("student.txt", "a") as file:
    file.write("Grade: A\n")

print("New data added.")
with open("student.txt", "r") as file:
    data = file.read()
print(data)
```

```python
# Pandas is a Python library mainly used for working with tabular data
# such as Excel sheets, CSV files, and database-like tables.
import pandas as pd

# Creating data
data = {
    "Name": ["Aman", "Riya", "Rahul", "Priya"],
    "Age": [20, 21, 19, 22],
    "Marks": [85, 92, 78, 88]
}

# Creating DataFrame
df = pd.DataFrame(data)

print(df)
```

```python
print(df[["Name", "Marks"]])
```

```python
average = df["Marks"].mean()

print("Average Marks:", average)
```

```python
print("Highest Marks:", df["Marks"].max())
```

```python
# Matplotlib is used to create graphs and charts in Python.
import matplotlib.pyplot as plt

tests = [1, 2, 3, 4, 5]
marks = [65, 72, 78, 85, 90]

plt.plot(tests, marks)

plt.xlabel("Test Number")
plt.ylabel("Marks")
plt.title("Student Performance")

plt.show()
```

```python
import matplotlib.pyplot as plt

tests = [1, 2, 3, 4, 5]
marks = [65, 72, 78, 85, 90]

plt.plot(tests, marks, marker="o")

plt.xlabel("Test Number")
plt.ylabel("Marks")
plt.title("Student Performance")

plt.show()
```

```python
import matplotlib.pyplot as plt

students = ["Aman", "Riya", "Rahul", "Priya"]
marks = [85, 92, 78, 88]

plt.bar(students, marks)

plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student Marks")

plt.show()
```

```python
import matplotlib.pyplot as plt

languages = ["Python", "Java", "C++", "C"]
students = [40, 30, 20, 10]

plt.pie(students, labels=languages, autopct="%1.1f%%")

plt.title("Programming Language Preference")

plt.show()
```

```python
import matplotlib.pyplot as plt

study_hours = [1, 2, 3, 4, 5, 6]
marks = [45, 50, 60, 68, 78, 90]

plt.scatter(study_hours, marks)

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")

plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Read CSV file using Pandas
df = pd.read_csv("students.csv")

# Step 2: Display DataFrame
print("Student Data:")
print(df)

# Step 3: Use NumPy for numerical calculations
marks = np.array(df["Marks"])

print("\nAverage Marks:", np.mean(marks))
print("Highest Marks:", np.max(marks))
print("Lowest Marks:", np.min(marks))

# Step 4: Visualize the data
plt.bar(df["Name"], df["Marks"])

plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student Marks")

plt.show()
```
